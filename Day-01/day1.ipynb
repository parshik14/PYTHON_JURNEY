```python
# one functions
num1 = 25
num2 = 4

operations = {
    "addition" : num1 + num2,
    "substarction" : num1- num2 ,
    "multiplication" : num1*num2,
    "divide" : num1/num2
}

for key,value in operations.items():
    if isinstance(value, float):
        print(f"{key}: {value:.2f}")
    else:
        print(f"{key}: {value}")
    
```
