```python
import math

# Constants
print(math.pi)      # 3.141592653589793
print(math.e)       # 2.718281828459045
print(math.inf)     # Infinity
print(math.nan)     # Not a Number

# Basic operations
math.sqrt(16)       # 4.0
math.pow(2, 3)      # 8.0 (2^3)
math.factorial(5)   # 120 (5!)
math.gcd(48, 18)    # 6 (greatest common divisor)

# Rounding
math.ceil(4.3)      # 5 (round up)
math.floor(4.7)     # 4 (round down)
math.trunc(4.7)     # 4 (remove decimal)

# Trigonometry (radians)
math.sin(math.pi / 2)   # 1.0
math.cos(0)             # 1.0
math.tan(math.pi / 4)   # 1.0

# Convert degrees/radians
math.radians(180)   # 3.141592653589793
math.degrees(math.pi)  # 180.0

# Logarithms
math.log(100, 10)   # 2.0 (log base 10)
math.log(math.e)    # 1.0 (natural log)
math.log10(1000)    # 3.0

# Check values
math.isnan(float('nan'))  # True
math.isinf(math.inf)      # True
math.isfinite(100)   
```

```python
def outer():
    x = "outer"
    
    def inner():
        nonlocal x  # Modify variable from enclosing scope
        x = "modified by inner"
    
    print(f"Before: {x}")
    inner()
    print(f"After: {x}")

outer()
# Before: outer
# After: modified by inner
```

```python
import re
import random
import string
from typing import List, Dict, Optional, Union
```

```python
VERSION = "1.0.0"
VOWELS = "aeiouAEIOU"
CONSONANTS = "bcdfghjklmnpqrstvwxyzBCDFGHJKLMNPQRSTVWXYZ"
```

```python
def analyze_string(text: str,case_sensitive:bool= False) -> Dict[str,Union[int,float,Dict]]:
    if not isinstance(text, str):
        raise TypeError("Input must be a string")
    
    if not text.strip():
        return {
            "length": 0,
            "words": 0,
            "sentences": 0,
            "characters": 0,
            "vowels": 0,
            "consonants": 0,
            "digits": 0,
            "spaces": 0,
            "special_chars": 0,
            "char_frequency": {},
            "word_frequency": {},
            "average_word_length": 0,
            "longest_word": "",
            "shortest_word": ""
        }
    
    # Normalize case if needed
    analysis_text = text if case_sensitive else text.lower()
    
    # Basic counts
    length = len(text)
    spaces = text.count(" ")
    characters = sum(1 for c in text if not c.isspace())
    
    # Count character types
    vowels = sum(1 for c in text if c in VOWELS)
    consonants = sum(1 for c in text if c in CONSONANTS)
    digits = sum(1 for c in text if c.isdigit())
    special_chars = sum(1 for c in text if not c.isalnum() and not c.isspace())
    
    # Sentence count (approximate - counts . ! ?)
    sentences = sum(1 for c in text if c in '.!?')
    sentences = max(sentences, 1)  # At least 1 sentence if text exists
    
    # Character frequency
    char_frequency = {}
    for char in analysis_text:
        if not char.isspace():
            char_frequency[char] = char_frequency.get(char, 0) + 1
    
    # Word analysis
    words = text.split()
    word_count = len(words)
    
    # Word frequency (case-insensitive by default)
    word_frequency = {}
    for word in words:
        clean_word = ''.join(c for c in word if c.isalnum())
        if clean_word:
            key = clean_word if case_sensitive else clean_word.lower()
            word_frequency[key] = word_frequency.get(key, 0) + 1
    
    # Word length analysis
    word_lengths = [len(w) for w in words if w]
    average_word_length = sum(word_lengths) / len(word_lengths) if word_lengths else 0
    longest_word = max(words, key=len) if words else ""
    shortest_word = min(words, key=len) if words else ""
    
    return {
        "length": length,
        "words": word_count,
        "sentences": sentences,
        "characters": characters,
        "vowels": vowels,
        "consonants": consonants,
        "digits": digits,
        "spaces": spaces,
        "special_chars": special_chars,
        "char_frequency": dict(sorted(char_frequency.items(), key=lambda x: x[1], reverse=True)),
        "word_frequency": dict(sorted(word_frequency.items(), key=lambda x: x[1], reverse=True)),
        "average_word_length": round(average_word_length, 2),
        "longest_word": longest_word,
        "shortest_word": shortest_word
    }
```

```python
def generate_password(length: int = 12,include_uppercase: bool = True,include_lowercase: bool = True,include_digits: bool = True,include_special: bool = True,exclude_ambiguous: bool = True) -> str:
    if length < 4:
        raise ValueError("Password length must be at least 4")
    
    # Build character pool
    char_pool = ""
    required_chars = []
    
    if include_lowercase:
        lowercase = string.ascii_lowercase
        if exclude_ambiguous:
            lowercase = lowercase.replace('l', '')
        char_pool += lowercase
        required_chars.append(random.choice(lowercase))
    
    if include_uppercase:
        uppercase = string.ascii_uppercase
        if exclude_ambiguous:
            uppercase = uppercase.replace('O', '').replace('I', '')
        char_pool += uppercase
        required_chars.append(random.choice(uppercase))
    
    if include_digits:
        digits = string.digits
        if exclude_ambiguous:
            digits = digits.replace('0', '').replace('1', '')
        char_pool += digits
        required_chars.append(random.choice(digits))
    
    if include_special:
        special = "!@#$%^&*()-_=+[]{}|;:,.<>?"
        char_pool += special
        required_chars.append(random.choice(special))
    
    if not char_pool:
        raise ValueError("At least one character type must be selected")
    
    # Generate password ensuring at least one of each required type
    password_list = required_chars.copy()
    remaining_length = length - len(required_chars)
    
    for _ in range(remaining_length):
        password_list.append(random.choice(char_pool))
    
    # Shuffle to avoid predictable pattern
    random.shuffle(password_list)
    
    return ''.join(password_list)
```
