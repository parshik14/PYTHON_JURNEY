<h1> Loops & Comprehensions</h1>

```python
fruits = ["A","B","C"]
integers = [4,2,3]
strings = "python"
objects = {"name":"parshik","surname":"chovatiya"}
```

```python
# only index
for i in range(len(integers)):
    print(i)

# only value
for i in integers:
    print(i)
    
# get both index and value
for i in range(len(integers)):
    print(f"index is {i} and values is {integers[i]}")
```

```python
for i in integers:
    print(i)
```

```python
for char in "python":
    print(char)
```

```python
for i in range(len(strings)):
    print(i)
```

```python
for i,j in objects.values():
    print(i)
```

```python
# parellel iteration 
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
cities = ["NYC", "LA", "Chicago"]

for name, age, city in zip(names, ages, cities):
    print(f"{name}, {age}, from {city}")
```

```python
# create dictionary from two list 
keys =["name","age"]
values = ["parshik",22]

person = dict(zip(keys,values))
person
```
