```python
# filter duplicates while maintaining order
def unique_ordered(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

unique_ordered([2,1, 2, 2, 3, 1, 4])
```
